```python
import logging
import requests

for symbol in SYMBOLS:
    try:
        logging.info("Canceling open orders symbol=%s", symbol)
        canceled_orders = Client.cancel_all_open_orders(symbol=symbol)
        logging.info("Canceled open orders symbol=%s result=%s", symbol, canceled_orders)
    except requests.HTTPError as exc:
        response = exc.response
        logging.exception(
            "Cancel all open orders failed symbol=%s status=%s response=%s",
            symbol,
            response.status_code if response is not None else None,
            response.text if response is not None else None,
        )
```
